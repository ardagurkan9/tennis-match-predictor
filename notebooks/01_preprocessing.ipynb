# ATP Match Preprocessing

Bu notebookta simdilik yalnizca ham ATP mac verileri yuklenir ve mac bittikten sonra belli olan, modelde kullanilmayacak kolonlar kaldirilir.

```python
from pathlib import Path

import pandas as pd

pd.set_option("display.max_columns", 120)
```

```python
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
RAW_DIR = PROJECT_ROOT / "data" / "raw"

raw_files = sorted(RAW_DIR.glob("atp_matches_*.csv"))
raw_files
```

```python
frames = [pd.read_csv(path) for path in raw_files]
matches = pd.concat(frames, ignore_index=True)

matches.shape
```

## Mac Sonrasi Belirlenen Kolonlari Kaldirma

```python
post_match_columns = [
    "score",
    "minutes",
    "w_ace",
    "w_df",
    "w_svpt",
    "w_1stIn",
    "w_1stWon",
    "w_2ndWon",
    "w_SvGms",
    "w_bpSaved",
    "w_bpFaced",
    "l_ace",
    "l_df",
    "l_svpt",
    "l_1stIn",
    "l_1stWon",
    "l_2ndWon",
    "l_SvGms",
    "l_bpSaved",
    "l_bpFaced",
]

post_match_columns
```

```python
matches_without_post_match_data = matches.drop(columns=post_match_columns, errors="ignore")

matches_without_post_match_data.head()
```

## Mac Oncesi Feature Adaylari

```python
matches_without_post_match_data["rank_diff"] = matches_without_post_match_data["winner_rank"] - matches_without_post_match_data["loser_rank"]
matches_without_post_match_data["age_diff"] = matches_without_post_match_data["winner_age"] - matches_without_post_match_data["loser_age"]
matches_without_post_match_data["height_diff"] = matches_without_post_match_data["winner_ht"] - matches_without_post_match_data["loser_ht"]
matches_without_post_match_data["rank_points_diff"] = matches_without_post_match_data["winner_rank_points"] - matches_without_post_match_data["loser_rank_points"]
matches_without_post_match_data["hand_matchup"] = matches_without_post_match_data["winner_hand"].fillna("U") + "_vs_" + matches_without_post_match_data["loser_hand"].fillna("U")

matches_without_post_match_data[["rank_diff", "age_diff", "height_diff", "rank_points_diff", "hand_matchup"]].head()
```
